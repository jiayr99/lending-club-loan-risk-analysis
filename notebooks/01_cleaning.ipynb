```python
# load the raw data
path = "/content/drive/MyDrive/lending_club/archive/loan.csv"
```

```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", 60)   # show more columns when printing
```

```python
# print columns
all_cols = pd.read_csv(path, nrows=5).columns.tolist()
print(len(all_cols))
print(all_cols)
```

```python
# keep relevant columns
keep_cols = [
    "id", "loan_amnt", "funded_amnt", "term", "int_rate", "installment",
    "grade", "sub_grade", "emp_length", "home_ownership", "annual_inc",
    "verification_status", "issue_d", "loan_status", "purpose",
    "addr_state", "dti", "total_rec_prncp", "total_rec_int"
]

# check column names
missing_cols = [c for c in keep_cols if c not in all_cols]
print("Not found:", missing_cols)

keep_cols = [c for c in keep_cols if c in all_cols]
```

```python
df = pd.read_csv(path, usecols=keep_cols, low_memory=False)
print(df.shape)   # (rows, columns)
```

```python
# share of missing values per column
df.isna().mean().sort_values(ascending=False)
```

**Observation**: `id` is 100% missing in this file version, so it cannot identify loans. I will drop it and create `loan_key` as a row number instead. `emp_length` is missing in about 6.5% of rows, so I will fill it with “Unknown” to keep those loans. `dti` and `annual_inc` have negligible missing values, which I will handle in the outlier step.

```python
# how many loans per status
df["loan_status"].value_counts()
```

```python
df["grade"].value_counts()
```

```python
df["purpose"].value_counts()
```

```python
# only keep loans with a known outcome
good = ["Fully Paid"]
bad = ["Charged Off", "Default"]

rows_before = len(df)
df = df[df["loan_status"].isin(good + bad)].copy()
print("Kept resolved loans:", len(df), "of", rows_before)

df["is_default"] = df["loan_status"].isin(bad).astype(int)
```

```python
# fix term to integer data type
df["term"] = df["term"].astype(str).str.extract(r"(\d+)")[0].astype(int)
```

```python
# fix text date to a real date
df["issue_d"] = pd.to_datetime(df["issue_d"], format="%b-%Y")
df["issue_year"] = df["issue_d"].dt.year
```

```python
# remove stray spaces for text
for col in ["grade", "sub_grade", "purpose", "home_ownership", "addr_state", "verification_status"]:
    df[col] = df[col].astype(str).str.strip()
```

```python
# drop id because all the rows are empty
df = df.drop(columns=["id"])
df = df.reset_index(drop=True)
df["loan_key"] = df.index + 1
```

```python
df["emp_length"] = df["emp_length"].fillna("Unknown")
```

```python
# check duplicates
dup_cols = ["loan_amnt", "term", "int_rate", "grade", "issue_d",
            "annual_inc", "addr_state", "purpose", "dti"]
print("Fully duplicated rows:", df.duplicated(subset=dup_cols).sum())
```

```python
dupes = df[df.duplicated(subset=dup_cols, keep=False)].sort_values(dup_cols)
dupes
```

**Observation**: Found 4 rows matching on 9 key columns. Compared all columns: they differed on `emp_length`. Decision: kept them, because they are not exact duplicates. Impact on results: negligible.

```python
# for outliers, look at extremes
df[["loan_amnt", "annual_inc", "dti", "int_rate"]].describe(percentiles=[0.01, 0.5, 0.99])
```

```python
# since dti is a ratio, negative values or less than 100 usually mean bad data. Income of zero or less is also suspicious
rows_before = len(df)
df = df[(df["dti"].isna()) | ((df["dti"] >= 0) & (df["dti"] <= 100))]
df = df[df["annual_inc"] > 0]
print("Removed suspicious rows:", rows_before - len(df))
```

```python
# sanity check
print(df.shape)
print("Overall default rate: ", round(df["is_default"].mean() * 100, 2), "%")
print(df.groupby("grade")["is_default"].mean().round(3))
print(df.dtypes)
```

```python
out_path = "/content/drive/MyDrive/lending_club/archive/loans_clean.csv"
df.to_csv(out_path, index=False)
print("Size (MB):", round(os.path.getsize(out_path) / 1e6, 1))
```
